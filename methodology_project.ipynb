```python
import re

def clean_column_name(col_name):
    # Remove leading/trailing whitespace
    col_name = col_name.strip()
    # Remove trailing question mark (؟) and non-breaking space (&nbsp;)
    col_name = col_name.replace('؟', '').replace('&nbsp;', '').strip()
    # Remove common Arabic interrogative prefixes
    if col_name.startswith('هل '):
        col_name = col_name[len('هل '):].strip()
    elif col_name.startswith('فاي '):
        col_name = col_name[len('فاي '):].strip()

    # Remove content within parentheses, as these are often explanations
    col_name = re.sub(r'\(.*\)', '', col_name).strip()

    return col_name

# Apply the cleaning function to all column names in the 'clean' DataFrame
new_column_names = {col: clean_column_name(col) for col in clean.columns}
clean_renamed = clean.rename(columns=new_column_names)

print("Original number of columns:", len(clean.columns))
print("Cleaned column names:")
for col in clean_renamed.columns:
    print(col)
```

```python
import numpy as np

# Make a copy to work with, in case we need to revert
temp_df = clean_renamed.copy()

# Create a mapping for renaming
rename_map = {}
for col in temp_df.columns:
    if 'Unnamed:' in col:
        # Get the value from the first row of this 'Unnamed' column
        potential_new_name = temp_df.at[temp_df.index[0], col]

        # Check if the potential new name is not NaN and is a string, then strip whitespace
        if pd.notna(potential_new_name) and isinstance(potential_new_name, str):
            potential_new_name_str = potential_new_name.strip()
            if potential_new_name_str:
                rename_map[col] = potential_new_name_str
            else:
                rename_map[col] = col # Keep original if after strip it's empty
        else:
            rename_map[col] = col # Keep original if it's NaN or not a string
    else:
        rename_map[col] = col # Keep existing good names

# Apply the renaming for 'Unnamed' columns
temp_df = temp_df.rename(columns=rename_map)

# Drop the first row, as it now contains redundant header information
clean_final = temp_df.drop(temp_df.index[0]).reset_index(drop=True)

print("\nFinal cleaned column names after using first row for 'Unnamed' columns:")
for col in clean_final.columns:
    print(col)

display(clean_final.head())
```

```python
for col in clean_final.columns:
    try:
        numeric_col = pd.to_numeric(clean_final[col], errors='coerce')

        if col == 'Gender':
            unique_gender_values = numeric_col.dropna().unique()

            if np.all(np.isin(unique_gender_values, [1, 2])):
                # Map 1 to 0 (False) and 2 to 1 (True) for boolean conversion
                mapping_gender = {1: 0, 2: 1}
                clean_final[col] = numeric_col.map(mapping_gender).astype('boolean')
            elif np.all(np.isin(unique_gender_values, [0, 1])):
                # If Gender already has 0s and 1s, convert directly to boolean
                clean_final[col] = numeric_col.astype('boolean')
            else:
                # If Gender has other unexpected values, keep as Int64 and print warning
                print(f"Warning: 'Gender' column contains unexpected values ({unique_gender_values}). Keeping as Int64.")
                clean_final[col] = numeric_col.astype('Int64')
        else:
            # Existing logic for other columns
            unique_values = numeric_col.dropna().unique()
            if np.all(np.isin(unique_values, [0, 1])):
                clean_final[col] = numeric_col.astype('boolean')
            else:
                clean_final[col] = numeric_col.astype('Int64')

    except Exception as e:
        print(f"Warning: Could not convert column '{col}'. Error: {e}. Keeping original dtype.")

print("\nDataFrame dtypes after conversion:")
print(clean_final.dtypes)
clean_final.head()
```

```python
import pandas as pd
import numpy as np

# Dictionary to store column groups and their binning rules
binning_rules = {
    'discrete_vars': [
        'General_Search_Ranking',
        'Search engines e.g. google',
        'Social media',
        'Prefer_Reviews',
        'RealPeople_Info_Acceptance',
        'Verify_Info_General',
        'Study_Level',
        'Google_Searches_out_of_5'
    ],
    'custom_bins': {
        'AI_Satisfaction_Rating': [0, 20, 40, 60, 80, 100],
        'SocialMedia_Satisfaction_Rating': [0, 20, 40, 60, 80, 100],
        'SearchEngine_Satisfaction_Rating': [0, 20, 40, 60, 80, 100]
    }
}

print("Generating Frequency Tables for quantitative variables:")

# Process discrete variables by directly counting values
for col_name in binning_rules['discrete_vars']:
    if col_name in final_data.columns:
        print(f"\n--- Frequency Table for discrete variable '{col_name}' ---")
        try:
            value_counts = final_data[col_name].value_counts().sort_index().to_frame(name='Count')
            value_counts['Percentage'] = (value_counts['Count'] / len(final_data)) * 100
            display(value_counts)
        except Exception as e:
            print(f"Could not create frequency table for '{col_name}'. Error: {e}.")
    else:
        print(f"Warning: Column '{col_name}' not found in final_data.")

# Process columns requiring custom bins (e.g., satisfaction ratings)
for col_name, bins_list in binning_rules['custom_bins'].items():
    if col_name in final_data.columns:
        print(f"\n--- Frequency Table for '{col_name}' (Custom Bins: {bins_list}) ---")
        try:
            binned_series = pd.cut(final_data[col_name].astype(float), bins=bins_list, include_lowest=True, right=True)
            frequency_table = binned_series.value_counts().sort_index().to_frame(name='Count')
            frequency_table['Percentage'] = (frequency_table['Count'] / len(final_data)) * 100
            display(frequency_table)
        except Exception as e:
            print(f"Could not create custom bins for '{col_name}'. Error: {e}. Showing value counts instead.")
            value_counts = final_data[col_name].value_counts().sort_index().to_frame(name='Count')
            value_counts['Percentage'] = (value_counts['Count'] / len(final_data)) * 100
            display(value_counts)
    else:
        print(f"Warning: Column '{col_name}' not found in final_data.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create a copy of final_data to apply string mappings for plotting
final_data_names = final_data.copy()

# Define mapping rules for different groups of boolean columns
mapping_rules = {
    'Yes/No_0_1': {
        True: 'Yes',
        False: 'No' # Assuming 1 was 'Yes' and 0 was 'No' after initial conversion
    },
    'Secure/Not Secure': {
        True: 'Secure',
        False: 'Not Secure'
    },
    'Used/Not Used': {
        True: 'Used',
        False: 'Not Used'
    },
    'Yes/No_True_False': {
        True: 'Yes',
        False: 'No'
    },
    'Male/Female': {
        True: 'Female',
        False: 'Male' # Based on previous mapping {1:0, 2:1} where 1 is Male, 2 is Female
    }
}

# Group columns by their intended mapping
columns_to_map = {
    'Yes/No_0_1': [
        'AI_usage_frequency',
        'SocialMedia_usage_frequency',
        'SearchEngine_usage_frequency'
    ],
    'Secure/Not Secure': [
        'secure ai',
        'secure search engines',
        'secure social media'
    ],
    'Used/Not Used': [
        'SocialMedia_for_Research_instagram', 'Tiktok', 'Youtube', 'X(twitter)', 'Facebook', 'Reddit'
    ],
    'Yes/No_True_False': [
        'No_SocialMedia_for_Research', 'SocialMedia_Research_speed', 'Preference_Security',
        'Preference_Info_Trust', 'Preference_Info_Quantity', 'Tiktok_Short_Info_Reviews',
        'No_SocialMedia_Usage'
    ],
    'Male/Female': ['Gender']
}

# Apply mappings to create new columns with string labels
print("Applying string labels to boolean columns for plotting...")
for map_type, cols in columns_to_map.items():
    for col in cols:
        if col in final_data_names.columns:
            # Ensure the column is boolean before applying the map
            if pd.api.types.is_bool_dtype(final_data_names[col]):
                final_data_names[f'{col}_label'] = final_data_names[col].map(mapping_rules[map_type])
            else:
                print(f"Warning: Column '{col}' is not boolean. Skipping string mapping.")
        else:
            print(f"Warning: Column '{col}' not found in final_data_names.")


print("Generating Bar Charts for boolean variables with custom labels:")

# Generate Bar Charts for the newly mapped categorical columns
for map_type, cols in columns_to_map.items():
    for col in cols:
        if f'{col}_label' in final_data_names.columns:
            try:
                value_counts = final_data_names[f'{col}_label'].value_counts().sort_index()
                plt.figure(figsize=(8, 5))
                value_counts.plot(kind='bar', color=['skyblue', 'lightcoral'])
                plt.title(f'Distribution of {col}')
                plt.xlabel(f'{col} (Label)')
                plt.ylabel('Count')
                plt.xticks(rotation=0) # Keep labels horizontal for clarity
                plt.tight_layout()
                plt.show()
            except Exception as e:
                print(f"Could not generate bar chart for '{col}_label'. Error: {e}.")
```
